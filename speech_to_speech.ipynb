```python
english_model_name = "facebook/mms-tts-eng"
yoruba_model_name = "facebook/mms-tts-yor"

english_tokenizer = AutoTokenizer.from_pretrained(english_model_name)
english_model = VitsModel.from_pretrained(english_model_name)

yoruba_tokenizer = AutoTokenizer.from_pretrained(yoruba_model_name)
yoruba_model = VitsModel.from_pretrained(yoruba_model_name)
def generate_speech(text, language, filename):

    if language == "english":
        tokenizer = english_tokenizer
        model = english_model

    elif language == "yoruba":
        tokenizer = yoruba_tokenizer
        model = yoruba_model

    else:
        raise ValueError("Language must be 'english' or 'yoruba'")

    inputs = tokenizer(text, return_tensors="pt")

    with torch.no_grad():
        output = model(**inputs).waveform

    audio = output.squeeze().cpu().numpy()

    scipy.io.wavfile.write(
        filename,
        rate=model.config.sampling_rate,
        data=audio
    )

    return filename
```

```python
english_file = generate_speech(
    "Good morning. How are you?",
    "english",
    "english_test.wav"
)
```
